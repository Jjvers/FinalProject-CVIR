## Preview Output Files
Kamu bisa melihat gambar/grafik *Live Training* dari model yang sedang berjalan. Stop Code Block di atas, jalankan command di bawah ini dan kamu bisa lihat progres grafiknya:

```
from IPython.display import Image, display
import glob
import os

image_paths = glob.glob('training_results/*.png')
if not image_paths:
    print('Belum ada grafik yang dihasilkan.')
for img_path in sorted(image_paths):
    print(f'\nMenampilkan: {os.path.basename(img_path)}')
    display(Image(filename=img_path))
```

```
# Download Model
from google.colab import files
if os.path.exists('trained_models/emotion_model.keras'):
    files.download('trained_models/emotion_model.keras')
else:
    print('Model final belum dikompilasi. Kita download versi checkpoint terbaru:')
    for model_path in glob.glob('trained_models/*.keras'):
        files.download(model_path)
```
